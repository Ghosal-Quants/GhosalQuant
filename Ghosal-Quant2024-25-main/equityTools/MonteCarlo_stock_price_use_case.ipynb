# Example usage of Monte Carlo Simulation

Here is an example notebook where I am making use of the Monte Carlo simulation script. The code isnt thoroughly commented, so probably easiest to see it used in practice.

```python
from MonteCarlo import MonteCarloSim
```

Packages needed inside of MonteCarlo are Pandas, numpy, scipy, matplotlib, datetime, and yfinance.

As default, for any ticker, you can simulate 90 days into the future using data recovered from the last 365 days to fit the distribution of daily changes. This runs for 1000 simulations.

```python
standard_MCmodel = MonteCarloSim('IMI.L')
minima_quantiles, end_point_quantiles = standard_MCmodel.plots_and_percentiles()
```

Note, the flat regions are for the weekends, where price is assumed to not vary at all. I'm sure this is an assumption that could be refined. 

The average return over the simulations, seen here in black, is the same as the average return over the time period the distribution is fitted over. As standard, this is 365 days. 

```python
minima_quantiles
```

## Arguments of MonteCarloSim

```python
# There are a number of tunable parameters 
tuned_MCmodel = MonteCarloSim(
    ticker = 'IMI.L',         # Any ticker as it appears on yahoo finance 
    start = '2018-01-01',     # Default start 1yr ago, begins simulation on current day
    end_sim = '2024-12-31',   # Default 90 day forecast 
    num_sims = 1000,          # Number of Monte Carlos to run
    num_plotted_sims = 100,   # Number of monte carlo paths shown on final plot
    show_fitted_dist = True,  # Show how the distribution of daily moves is parametrised, good for a visual goodness of fit test 
    save_plot = False)        # Saves a high dpi png to directory 
minima_quantiles, end_point_quantiles = tuned_MCmodel.plots_and_percentiles(sim_color='black', avg_color='green', sim_alpha=0.05)
```
